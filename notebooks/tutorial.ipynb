```python
# 1. Quick Setup
# Import all necessary components
from scientific_template import *
import pandas as pd
import numpy as np

# Setup logging
logger = setup_logging(level="INFO")

# Set plotting style
set_style(style="seaborn-v0_8", context="notebook")

logger.info("Setup complete!")
```

## 2. Data Loading

Load data from various formats (CSV, Excel, JSON, Parquet) using DataLoader.

```python
# Load data with DataLoader
loader = DataLoader(engine="pandas")
df = loader.load_csv("data/input/sales_data.csv")

# Quick data overview
print("Data Shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())

# Get data info
processor = DataProcessor()
info = processor.get_info(df)
print("\nData Info:")
print(info)
```

## 3. Data Processing

Clean and preprocess data using DataCleaner and DataProcessor.

```python
# Data cleaning
cleaner = DataCleaner()
df_clean = cleaner.standardize_column_names(df)
df_clean = cleaner.fix_data_types(df_clean)

# Data processing
processor = DataProcessor()
df_processed = processor.remove_duplicates(df_clean)
df_processed = processor.handle_missing(df_processed, strategy="fill_mean")

# Data transformation
transformer = DataTransformer()
df_transformed = transformer.encode_categorical(df_processed, columns=["category", "region"])

print("Processed Data Shape:", df_transformed.shape)
display(df_transformed.head())
```

## 4. Statistical Analysis

Perform statistical analysis using StatisticalAnalyzer.

```python
# Initialize analyzer
analyzer = StatisticalAnalyzer()

# Descriptive statistics
desc_stats = analyzer.descriptive_stats(df)
print("Descriptive Statistics:")
display(desc_stats)

# Correlation analysis
corr_matrix = analyzer.correlation(df.select_dtypes(include=[np.number]))
print("\nCorrelation Matrix:")
display(corr_matrix)

# Pearson correlation between sales and quantity
corr_result = analyzer.pearson_correlation(df["sales"], df["quantity"])
print(f"\nPearson Correlation (sales vs quantity):")
print(f"  r = {corr_result['correlation']:.3f}")
print(f"  p-value = {corr_result['p_value']:.4f}")
```

## 5. Visualization

Create publication-quality visualizations using Plotter.

```python
# Initialize plotter
plotter = Plotter(style="seaborn-v0_8", context="notebook")

# Create histogram
fig1 = plotter.create_histogram(df, column="sales", title="Sales Distribution")

# Create bar plot
fig2 = plotter.create_bar_plot(df, x="category", y="sales", title="Sales by Category")

# Create scatter plot
fig3 = plotter.create_scatter_plot(df, x="quantity", y="sales", 
                                   hue="category", title="Sales vs Quantity")

# Create correlation matrix heatmap
fig4 = plotter.create_correlation_matrix(df.select_dtypes(include=[np.number]))

# Save figures
plotter.save_figure(fig1, "output/figures/sales_histogram.png")
plotter.save_figure(fig2, "output/figures/sales_barplot.png")
plotter.save_figure(fig3, "output/figures/sales_scatter.png")
plotter.save_figure(fig4, "output/figures/correlation_heatmap.png")

print("Visualizations saved to output/figures/")
```

## 6. Machine Learning

Build and evaluate machine learning models using ModelPipeline and CrossValidator.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression

# Prepare data for ML
X = df_transformed.drop(["sales", "date"], axis=1)
y = df_transformed["sales"]

# Train model with ModelPipeline
pipeline = ModelPipeline(model=RandomForestRegressor(n_estimators=100))
results = pipeline.fit(X, y, test_size=0.2, random_state=42)

print("Model Training Results:")
print(f"  R² Score: {results['metrics']['r2']:.4f}")
print(f"  RMSE: {results['metrics']['rmse']:.2f}")

# Cross-validation
cv = CrossValidator(cv=5)
cv_results = cv.validate(
    model=RandomForestRegressor(n_estimators=100),
    X=X, y=y,
    scoring="r2"
)

print("\nCross-Validation Results:")
print(f"  Mean R²: {cv_results['mean_score']:.4f}")
print(f"  Std: {cv_results['std_score']:.4f}")
print(f"  Range: [{cv_results['min_score']:.4f}, {cv_results['max_score']:.4f}]")
```
